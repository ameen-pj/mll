```python
# Loading & Preprocessing
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split

ds = load_wine()

X = ds.data
Y = ds.target

xtr, xte, ytr, yte = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import f1_score, accuracy_score, precision_score, recall_score

architectures = [(50, ), (50, 10, ), (100, 50, 20,)]
activations = ['tanh', 'relu', 'logistic']

for arch in architectures:
    for act in activations:
        mlpc = MLPClassifier(hidden_layer_sizes = arch, activation=act, max_iter=20000)
        mlpc.fit(xtr, ytr)
        labels = mlpc.predict(xte)

        print(f"-- {arch} with {act} --")
        print(f"Accuracy : {accuracy_score(yte, labels)}")
        print(f"f1 : {f1_score(yte, labels, average='weighted', zero_division=False)}")
        print(f"precision : {precision_score(yte, labels, average='weighted', zero_division=False)}")
        print(f"recall : {recall_score(yte, labels, average='weighted', zero_division=False)}")
```
