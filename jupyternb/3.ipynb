```python
import pandas as pd

ds = pd.read_csv("./datasets/diabetes.csv")
    
X = ds.iloc[:, :-1]
Y = ds.iloc[:, -1]
```

```python
from sklearn.model_selection import train_test_split

xtr, xte, ytr, yte = train_test_split(X, Y, test_size = 0.2)

from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

xtr = sc.fit_transform(xtr)
xte = sc.transform(xte)
```

```python
from sklearn.linear_model import LinearRegression, LassoCV, RidgeCV


lr = LinearRegression()
lr.fit(xtr, ytr)
y1 = lr.predict(xte)

alphas = [0.01, 0.1, 1, 10, 100, 1000]

lasso = LassoCV(alphas=alphas, cv=5, max_iter=10000)
lasso.fit(xtr, ytr)
y2 = lasso.predict(xte)


ridge = RidgeCV(alphas=alphas, cv=5)
ridge.fit(xtr, ytr)
y2 = ridge.predict(xte)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

print("Linear Regression:")
print("MSE: ", mean_squared_error(yte,y1))
print("R2 Score: ", r2_score(yte, y1))

print("Lasso Regression:")
print("MSE: ", mean_squared_error(yte, y2))
print("R2 Score: ", r2_score(yte, y2))

print("Ridge Regression:")
print("MSE: ", mean_squared_error(yte, y2))
print("R2 Score: ", r2_score(yte, y2))
```
