```python
from sklearn.datasets import fetch_20newsgroups

ds = fetch_20newsgroups()

X = ds.data
Y = ds.target
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

xtr, xte, ytr, yte = train_test_split(X, Y, test_size=0.2)

cv = CountVectorizer()
xtr = cv.fit_transform(xtr)
xte = cv.transform(xte)

```

```python
from sklearn.naive_bayes import MultinomialNB, BernoulliNB

mnb = MultinomialNB()
mnb.fit(xtr, ytr)
l1 = mnb.predict(xte)

bnb = BernoulliNB()
bnb.fit(xtr, ytr)
l2 = bnb.predict(xte)

```

```python
from sklearn.metrics import accuracy_score, f1_score

print("Multinomial Naive Bayes")
print(f"Accuracy: {accuracy_score(yte, l1)}")
print(f"F1 Score: {f1_score(yte, l1, average='weighted')}")

print("\nBernoulli Naive Bayes")
print(f"Accuracy: {accuracy_score(yte, l2)}")
print(f"F1 Score: {f1_score(yte, l2, average='weighted')}")
```
