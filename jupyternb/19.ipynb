```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import BaggingClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
```

```python
ds = pd.read_csv("./datasets/titanic.csv")
ds
```

```python
ds = ds[["pclass", "sex", "age", "cabin", "survived"]].dropna()
ds
```

```python
X = ds.iloc[:, :-1]
Y = ds.iloc[:, -1]

le = LabelEncoder()

for col in X.columns:
    X[col] = le.fit_transform(X[col])
```

```python
# Splitting the dataset
xtr, xte, ytr, yte = train_test_split(X, Y, test_size = 0.2)
```

```python
# Method 1 - Bagging
bagging = BaggingClassifier(estimator=DecisionTreeClassifier(), n_estimators=10, bootstrap=True)
bagging.fit(xtr, ytr)

bagging_pred = bagging.predict(xte)
```

```python
# Method 2- Boosting
boosting = AdaBoostClassifier(estimator=DecisionTreeClassifier(), n_estimators=10)
boosting.fit(xtr, ytr)

boosting_pred = boosting.predict(xte)
```

```python
from sklearn.metrics import classification_report

print(classification_report(yte, bagging_pred))
print(classification_report(yte, boosting_pred))
```
