```python
# Loading the csv dataset
import pandas as pd

ds = pd.read_csv("./datasets/iris.csv")
ds = ds.dropna()
ds
```

```python
X = ds.iloc[:, :-1]
Y = ds.iloc[:, -1]

print(X)
print(Y)
```

```python
# Boostrapping
from sklearn.utils import resample
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import f1_score, accuracy_score
import numpy as np

acc_bs = []
f1_bs =  []

for i in range(30, 41):
    x_sample, y_sample = resample(X, Y, replace = True, random_state=i)

    model = DecisionTreeClassifier()
    model.fit(x_sample, y_sample)

    y_pred = model.predict(X)

    acc_bs.append(accuracy_score(Y, y_pred))
    f1_bs.append(f1_score(Y, y_pred, average='weighted'))


print(f"Accuracy : {np.mean(acc_bs) * 100} %")
print(f"F1 Score : {np.mean(f1_bs)}")
```

```python
# K-Fold Cross Validation
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score

kf = KFold(n_splits = 5, shuffle = True, random_state=42)

model = DecisionTreeClassifier()


acc_scores = cross_val_score(
    model, X, Y, cv=kf, scoring='accuracy'
)

f1_scores = cross_val_score(
    model, X, Y, cv=kf, scoring='f1_weighted'
)

print("\nK-Fold Cross Validation Results")
print("Average Accuracy:", np.mean(acc_scores) * 100)
print("Average F1 Score:", np.mean(f1_scores))
```
