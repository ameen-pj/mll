```python
# Data Loading & Feature, Target separation
import pandas as pd

ds = pd.read_csv("./datasets/fashion_mnist.csv")

X = ds.iloc[:, 1:]
Y = ds.iloc[:, [0]]
```

```python
# Spliting Training and Testing Dataset
from sklearn.model_selection import train_test_split

xtr, xte, ytr, yte = train_test_split(X, Y, test_size=0.2)
```

```python
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score, f1_score, precision_score

kernels = ['linear', 'poly', 'rbf']

for k in kernels:
    svc = SVC(kernel=k, max_iter=50)
    svc.fit(xtr, ytr)
    labels = svc.predict(xte)

    print(f"----- {k} -----")
    print(accuracy_score(yte, labels) * 100)
    print(f1_score(yte, labels, average='weighted'))
    print(precision_score(yte, labels, average='weighted'))
    # print(classification_report(yte, labels))
```
