```python
from sklearn.datasets import load_iris

ds = load_iris()
X = ds.data
Y = ds.target
```

```python
# Boostrapping
from sklearn.utils import resample
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, f1_score
import numpy as np

acc = []
f1 = []

for i in range(1, 30):
    x_sample, y_sample = resample(X, Y, replace=True, random_state=i)
    dtc = DecisionTreeClassifier()
    dtc.fit(x_sample, y_sample)

    predict = dtc.predict(X)
    acc.append(accuracy_score(Y, predict))
    f1.append(f1_score(Y, predict, average='weighted'))

print(f"Accuracy : {np.mean(acc)}")
print(f"F1 : {np.mean(f1)}")
```

```python
from sklearn.model_selection import KFold, cross_val_score

kf = KFold(n_splits=10, shuffle=True, random_state=42)
model = DecisionTreeClassifier()

acc = cross_val_score(estimator=model, X=X, y=Y, cv=kf, scoring='accuracy')
f1 = cross_val_score(estimator=model, X=X, y=Y, cv=kf, scoring='f1_weighted')

print(np.mean(acc))
print(np.mean(f1))
```
