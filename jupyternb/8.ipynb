```python
import pandas as pd

df = pd.read_csv("./datasets/fashion_mnist.csv")
df = df.dropna()

Y = df["label"]
X = df.iloc[:, 1:]
# Normalizing pixel values
X = X / 255.0
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

xtr, xte, ytr, yte = train_test_split(X, Y, random_state= 42, test_size= 0.2)

scaler = StandardScaler()

xtr = scaler.fit_transform(xtr)
xte = scaler.transform(xte)
```

```python
from sklearn.neighbors import KNeighborsClassifier 

kvals = [3, 5, 7, 9,11]
labels = []

for k in kvals:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(xtr, ytr)
    label = knn.predict(xte)
    labels.append(label)
```

```python
from sklearn.metrics import accuracy_score, f1_score

for label in labels:
    print(accuracy_score(yte, label) * 100)
    print(f1_score(yte, label, average='weighted'))
```
